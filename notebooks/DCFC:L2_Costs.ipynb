```python
import pandas as pd

# === CONFIGURATION ===
# Path to your Excel file (change this if needed)
file_path = "/Users/simon/git/CaliforniaElectricityCost/california-electrification-costs/AB2127_LoadCurveData_2ndAssessment_partial.xlsx"


# Constants
fleet_size = 7_100_000
minutes_to_hours = 60
MW_to_kW = 1000
price_public_work = 0.30  # $/kWh
price_dc_fast = 0.40      # $/kWh
days_per_year = 365

# === LOAD DATA ===
# Load the sheet "Figure 20", skip header rows
df = pd.read_excel(file_path, sheet_name="Figure 20", skiprows=2)

# Rename columns for easier access
df.columns = ["time", "Residential Level 1", "Residential Level 2", 
              "Public and Work Level 2", "DC Fast"]

# Keep only the first 1440 rows (24 hours, 1-minute intervals)
df_24h = df.iloc[:1440]

# === PROCESS DATA ===
# Sum the columns
public_work_l2_sum = df_24h["Public and Work Level 2"].sum()
dc_fast_sum = df_24h["DC Fast"].sum()

# Function to calculate annual cost per vehicle
def compute_annual_cost(total_mw_min, price_per_kwh):
    return ((total_mw_min / minutes_to_hours) / fleet_size) * MW_to_kW * price_per_kwh * days_per_year

# Calculate costs
public_work_cost = compute_annual_cost(public_work_l2_sum, price_public_work)
dc_fast_cost = compute_annual_cost(dc_fast_sum, price_dc_fast)

# === OUTPUT RESULTS ===
print(f"Public and Work Level 2: ${public_work_cost:.2f} per vehicle per year")
print(f"DC Fast: ${dc_fast_cost:.2f} per vehicle per year")
```
